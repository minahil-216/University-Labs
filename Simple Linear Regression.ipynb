```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Load the dataset
data = pd.read_csv(
    r"C:\Users\Cis\University-Labs\house_price_regression_dataset.csv"
)

# 2. Separate features and target
X = data.drop("House_Price", axis=1)
y = data["House_Price"]

# 3. First 200 houses = test data
X_test = X.iloc[:200]
y_test = y.iloc[:200]

# 4. Remaining houses = training data
X_train = X.iloc[200:]
y_train = y.iloc[200:]

# 5. Create the Linear Regression model
model = LinearRegression()

# 6. Train the model using the remaining houses
model.fit(X_train, y_train)

# 7. Predict prices of the first 200 houses
y_pred = model.predict(X_test)

# 8. Display actual and predicted prices
results = pd.DataFrame({
    "Actual_Price": y_test.values,
    "Predicted_Price": y_pred
})

print("\nFirst 200 House Price Predictions:")
print(results)

# 9. Model performance
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\nModel Performance:")
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Train on the historical data
data = pd.read_csv("house_price_regression_dataset.csv")
X = data.drop("House_Price", axis=1)
y = data["House_Price"]

model = LinearRegression()
model.fit(X, y)

# 2. Load the new houses (CSV with the same columns, minus House_Price)
new_houses = pd.read_csv("new_houses.csv")

# 3. Predict (reorder columns to match training)
new_houses["Predicted_Price"] = model.predict(new_houses[X.columns]).round(2)

# 4. Show and save
print(new_houses)
new_houses.to_csv("predicted_prices.csv", index=False)
```
